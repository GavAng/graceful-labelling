```python
import networkx as nx

from graphs import PathGraph, CycleGraph, CompleteGraph, StarGraph, SunGraph, WheelGraph
from graphs.tools import draw_graphs
from graphs.graceful_labellings import (
    find_graceful_labellings,
    find_alpha_valuations,
    find_eta_valuations,
)
from graphs.utils import Position, Vertex
```

```python
for n_vertices in range(2, 8):
    P_n = nx.Graph(PathGraph.edges(n_vertices))
    draw_graphs(find_alpha_valuations(P_n), layout=PathGraph.layout(n_vertices))
```

```python
rosa_star = [(1, 2), (2, 0), (3, 4), (4, 0), (5, 6), (6, 0)]

edge_sets = [
    CycleGraph.edges(4),
    rosa_star,
    PathGraph.edges(5),
]
layouts: list[dict[Vertex, Position] | None] = [
    CycleGraph.layout(4),
    None,
    PathGraph.layout(5),
]

for edges, layout in zip(edge_sets, layouts):
    print(f"Graph {edges}")
    graph = nx.Graph(edges)
    print("Graceful labellings:")
    try:
        draw_graphs(find_graceful_labellings(graph), layout=layout)
    except StopIteration:
        print("No graceful labellings found.")
    print("Alpha-valuations:")
    try:
        draw_graphs(find_alpha_valuations(graph), layout=layout)
    except StopIteration:
        print("No alpha-valuations found.")
```
